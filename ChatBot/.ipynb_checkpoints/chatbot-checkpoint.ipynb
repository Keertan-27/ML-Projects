```python
#importing the libraries
import numpy as np
import tensorflow as tf
import pandas as pd
import json
import random
import nltk
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.layers import Input, Embedding, LSTM, Dense, GlobalMaxPooling1D, Flatten
from tensorflow.keras.models import Model
import matplotlib.pyplot as plt

#Other
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_json('contents.json')
df.head()
```

```python
#importing the dataset
with open('contents.json') as intent:
    data1 = json.load(intent)
```

```python
# getting all the data into list
tags = []
patterns = []
responses = {}
for intent in data1['intents']:
    responses[intent['tag']] = intent['responses']
    for lines in intent['patterns']:
        patterns.append(lines)
        tags.append(intent['tag'])
```

```python
# converting into dataframe
data = pd.DataFrame({"patterns": patterns,
                    "tags": tags})
data
```

```python
# shuffiling the data
data = data.sample(frac = 1)
data
```

## Pre-Processing
Important preprocessing removing puncutations, lowercase the letters, encoding the textual data into numerical data 

```python
# removing puncutation 
import string
data['patterns'] = data['patterns'].apply(lambda wrd:[ltrs.lower() for ltrs in wrd if ltrs not in string.punctuation])
data['patterns'] = data['patterns'].apply(lambda wrd: ''.join(wrd))
data
```

```python
data.head()
```

```python
#tokenzine the data
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer(num_words = 2000)
tokenizer.fit_on_texts(data['patterns'])
train = tokenizer.texts_to_sequences(data['patterns'])
# apply padding 
from tensorflow.keras.preprocessing.sequence import pad_sequences
x_train = pad_sequences(train)

#encoding the outputs
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y_train = le.fit_transform(data['tags'])
```

Tensorflow's tokenizer assigns a unique token to each distinct word and padding is done to get all the data to the same length so as to send it to the RNN layer 

```python
input_shape = x_train.shape[1]
print(input_shape)
```

```python
# define vocabulary 
vocabulary = len(tokenizer.word_index)
print("number of unique words : ", vocabulary)
output_length = le.classes_.shape[0]
print("Output lenght : ", output_length)
```

## Neural Network
This network consists of an embeding layer which is one of the most powerful things in the field of natural language processing. The output of the embeding layer is the input of the recurent layer with LSTM gate. Then the output is flattend and a regular dense layer is used with a softmax activation function.

The main part is the embedding layer which gives has a corresponding vector for each word in dataset

```python
# creating the model 
i = Input(shape=(input_shape))
x = Embedding(vocabulary+1, 10)(i)
x = LSTM(10, return_sequences=True)(x)
x = Flatten()(x)
x= Dense(output_length, activation= "softmax")(x)
model = Model(i,x)
```

```python
# compiling the model 
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
# traning the model
train = model.fit(x_train, y_train, epochs = 300)
```

## Model Analysis

```python
#plotting model accuracy
plt.plot(train.history['accuracy'], label='training set accuracy')
plt.plot(train.history['loss'], label='traning set loss')
plt.legend()
```

## Testing the model
In this model we have used very small amount of data but it can be used with big data as well 

```python
#chatting
import random

while True:
    texts_p = []
    prediction_input = input("You: ")
    # removivng puncutations and converting to lowerecase
    prediction_input = [letters.lower() for letters in prediction_input if letters not in string.punctuation]
    prediction_input = ''.join(prediction_input)
    texts_p.append(prediction_input)
    
    #tokenizatioin and paddding 
    prediction_input = tokenizer.texts_to_sequences(texts_p)
    prediction_input = np.array(prediction_input).reshape(-1)
    prediction_input = pad_sequences([prediction_input],input_shape)
    
    
    #getting output from the model
    output = model.predict(prediction_input)
    output = output.argmax()
    
    #finding the right tag and predicting
    response_tag = le.inverse_transform([output])[0]
    print("ChatBot : ", random.choice(responses[response_tag]))
    if response_tag == 'goodbye':
        break
```
